```python
# Import Required Libraries
import sys
import os
from ipywidgets import VBox, Checkbox, Output, Label, HTML
from IPython.display import display, clear_output

# Add path for custom modules
sys.path.insert(0, '/home/crimsondeepdarshak/Desktop/Deep_Darshak/References/Build_1_docs/Fishing_areas')

from maritime_layers.utils.map_layers import (
    TerritorialSeaLayer,
    ContiguousZoneLayer,
    EEZLayer,
    FAOFishingAreaLayer,
    MaritimeMapController
)

# Base path
base_path = '/home/crimsondeepdarshak/Desktop/Deep_Darshak/References/Build_1_docs/Fishing_areas'

print("✓ All libraries imported successfully")
```

```python
# Initialize the Maritime Map Controller
print("\n" + "="*70)
print("INITIALIZING MARITIME MAP CONTROLLER")
print("="*70)

# Create map controller centered on world
map_controller = MaritimeMapController(center=[0, 0], zoom_start=2)

# Initialize layer objects
print("\nInitializing maritime layers...")

territorial_sea = TerritorialSeaLayer()
contiguous_zone = ContiguousZoneLayer()
eez = EEZLayer()
fao_areas = FAOFishingAreaLayer()

print("✓ Layer objects created")
```

```python
# Load Data for All Layers
print("\n" + "="*70)
print("LOADING GEOGRAPHIC DATA")
print("="*70)

# 12NM Territorial Sea
territorial_sea_path = os.path.join(
    base_path, 'World_Boundaries', 'World_12NM_v4_20231025_gpkg', 'eez_12nm_v4.gpkg'
)
territorial_sea.load_data(territorial_sea_path)

# 24NM Contiguous Zone
contiguous_zone_path = os.path.join(
    base_path, 'World_Boundaries', 'World_24NM_v4_20231025_gpkg', 'eez_24nm_v4.gpkg'
)
contiguous_zone.load_data(contiguous_zone_path)

# EEZ
eez_path = os.path.join(
    base_path, 'World_EEZ_v12_20231025_gpkg', 'eez_v12.gpkg'
)
eez.load_data(eez_path)

# FAO Fishing Areas
fao_path = os.path.join(base_path, 'FAO_AREAS_ERASE.json')
fao_areas.load_data(fao_path)

print("\n✓ All data loaded successfully")
```

```python
# Register Layers with Controller
print("\nRegistering layers with map controller...")

map_controller.add_layer(territorial_sea, active=False)
map_controller.add_layer(contiguous_zone, active=False)
map_controller.add_layer(eez, active=False)
map_controller.add_layer(fao_areas, active=False)

print("✓ All layers registered")
print(f"\nAvailable layers:")
for layer_name in map_controller.layers.keys():
    print(f"  - {layer_name}")
```

```python
# Create Interactive Layer Controls
print("\n" + "="*70)
print("CREATING INTERACTIVE CONTROLS")
print("="*70)

# Create output widget for messages
output_messages = Output()
map_output = Output()

# Create checkboxes for each layer
checkbox_12nm = Checkbox(
    value=False,
    description='12NM Territorial Sea',
    indent=False,
    style={'description_width': 'initial'}
)

checkbox_24nm = Checkbox(
    value=False,
    description='24NM Contiguous Zone',
    indent=False,
    style={'description_width': 'initial'}
)

checkbox_eez = Checkbox(
    value=False,
    description='Exclusive Economic Zone (EEZ)',
    indent=False,
    style={'description_width': 'initial'}
)

checkbox_fao = Checkbox(
    value=False,
    description='FAO Fishing Areas',
    indent=False,
    style={'description_width': 'initial'}
)

print("✓ Interactive controls created")
```

```python
# Define Callback Function to Update Map
def update_map(change=None):
    """Update map based on checkbox selections"""
    with map_output:
        clear_output(wait=True)
        
        # Determine which layers are selected
        selected_layers = []
        
        if checkbox_12nm.value:
            selected_layers.append('12NM Territorial Sea')
        if checkbox_24nm.value:
            selected_layers.append('24NM Contiguous Zone')
        if checkbox_eez.value:
            selected_layers.append('Exclusive Economic Zone (EEZ)')
        if checkbox_fao.value:
            selected_layers.append('FAO Fishing Areas')
        
        if selected_layers:
            # Create new map for rendering
            map_controller.render_layers(selected_layers)
            map_controller.add_legend()
            
            print(f"Rendering {len(selected_layers)} layer(s)...")
            for layer_name in selected_layers:
                print(f"  ✓ {layer_name}")
            
            # Display map
            display(map_controller.get_map())
        else:
            print("ℹ️  No layers selected. Please select at least one layer to visualize.")
            display(map_controller.get_map())

# Attach callbacks to checkboxes
checkbox_12nm.observe(update_map, names='value')
checkbox_24nm.observe(update_map, names='value')
checkbox_eez.observe(update_map, names='value')
checkbox_fao.observe(update_map, names='value')

print("✓ Callbacks attached to checkboxes")
```

```python
# Display Interactive Control Panel
print("\n" + "="*70)
print("INTERACTIVE MARITIME BOUNDARIES VISUALIZATION")
print("="*70)

# Create header
header = HTML("""
<div style="background-color: #f0f0f0; padding: 15px; border-radius: 5px; margin-bottom: 15px;">
    <h2 style="margin: 0; color: #333;">🗺️ Maritime Boundaries Explorer</h2>
    <p style="margin: 5px 0; color: #666; font-size: 14px;">
        Toggle the layers below to visualize different maritime boundaries. 
        Click on any zone in the map for detailed information.
    </p>
</div>
""")

# Create control panel
control_panel = VBox([
    header,
    Label('Select Maritime Boundaries to Display:'),
    checkbox_12nm,
    checkbox_24nm,
    checkbox_eez,
    checkbox_fao,
])

# Display control panel
display(control_panel)
display(map_output)

# Display initial map
update_map()
```

```python
# Save Combined Map
print("\n" + "="*70)
print("SAVING MAP")
print("="*70)

# Create new map with all layers enabled
full_map_controller = MaritimeMapController(center=[0, 0], zoom_start=2)
full_map_controller.add_layer(territorial_sea)
full_map_controller.add_layer(contiguous_zone)
full_map_controller.add_layer(eez)
full_map_controller.add_layer(fao_areas)

# Render all layers
full_map_controller.render_layers([
    '12NM Territorial Sea',
    '24NM Contiguous Zone',
    'Exclusive Economic Zone (EEZ)',
    'FAO Fishing Areas'
])

full_map_controller.add_legend()

# Save map
output_path = os.path.join(base_path, 'maritime_boundaries_interactive.html')
full_map_controller.save(output_path)

print(f"\n✓ Combined map with all layers saved to:")
print(f"  {output_path}")
```

```python
# Export Individual Layer GeoJSON Files
print("\n" + "="*70)
print("EXPORTING GEOJSON FILES")
print("="*70)

output_geojson_dir = os.path.join(base_path, 'maritime_layers', 'processed_data')
os.makedirs(output_geojson_dir, exist_ok=True)

# Export each layer
territorial_sea.to_geojson(os.path.join(output_geojson_dir, '12nm_territorial_sea.geojson'))
contiguous_zone.to_geojson(os.path.join(output_geojson_dir, '24nm_contiguous_zone.geojson'))
eez.to_geojson(os.path.join(output_geojson_dir, 'eez.geojson'))
fao_areas.to_geojson(os.path.join(output_geojson_dir, 'fao_fishing_areas.geojson'))

print(f"\n✓ All GeoJSON files exported to:")
print(f"  {output_geojson_dir}")
```
